### Semantic Scholar

A API REST do Semantic Scholar permite que você encontre e explore dados de publicações científicas sobre autores, artigos, citações, locais e muito mais. A API está organizada nos seguintes serviços:

- Academic Graph : fornece dados sobre autores, artigos, citações, locais, incorporações do SPECTER2 e muito mais, permitindo que você crie um link direto para a página correspondente em semanticscholar.org para obter mais informações.
- Recomendações : Fornece artigos recomendados semelhantes a um artigo específico.
- Conjuntos de dados : fornece links para download de conjuntos de dados no gráfico acadêmico.

Link: https://www.semanticscholar.org/product/api

```python
url = 'https://api.semanticscholar.org/graph/v1/paper/search'
params = {
    'query': 'deep learning music emotion',
    'limit': 5,
    'fields': 'title,authors,year,abstract,url'
}

response = requests.get(url, params=params)
data = response.json()

for paper in data['data']:
    print(f"\n{paper['title']} ({paper['year']})")
    print(f"Autores: {[a['name'] for a in paper['authors']]}")
    print(f"URL: {paper['url']}")
```

```python
paper
```

```python
list_results = list()
for paper in data['data']:
    titulo = paper["title"]
    link = paper["url"]
    abstract = paper["abstract"]
    # print(work["display_name"], "-", work.get("doi", "sem DOI"))
    if link is not None and titulo is not None and abstract is not None:
      list_results.append((titulo, link, abstract))
```

```python
list_results
```

```python
def limpar_unicode(texto):
    return unicodedata.normalize('NFKD', texto).encode('latin-1', 'ignore').decode('latin-1')
```

```python
pdf = FPDF()
pdf.set_auto_page_break(auto=True, margin=15)
pdf.add_page()
pdf.set_font("Arial", size=12)

for i, (title, link, abstract) in enumerate(list_results, 1):
    pdf.set_font("Arial", style="B", size=12)
    pdf.multi_cell(0, 10, f"{i}. {title}")
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, f"Link: {link}")
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, limpar_unicode(abstract))
    pdf.ln()

pdf.output("resumos_semanticscholar.pdf")
```

### ArXiv

O arXiv é um serviço de distribuição gratuita e um arquivo de acesso aberto com quase 2,4 milhões de artigos acadêmicos nas áreas de física, matemática, ciência da computação, biologia quantitativa, finanças quantitativas, estatística, engenharia elétrica e ciência de sistemas, e economia. Os materiais neste site não são revisados ​​por pares pelo arXiv.

Link: https://arxiv.org/   
Link da api: https://info.arxiv.org/help/api/basics.html

```python
search = arxiv.Search(
    query="(dengue OR 'dengue fever') AND (prediction OR forecasting)",
    max_results=5,
    sort_by=arxiv.SortCriterion.SubmittedDate,
)

print(search)
for result in search.results():
    print(result.title)
    print(result.entry_id)
    print(result.summary[:10], '\n')
```

```python
print(result)
```

```python
print(result.summary)
```

```python
list_results = list()
for result in search.results():
    titulo = result.title
    link = result.entry_id
    abstract = result.summary
    # print(work["display_name"], "-", work.get("doi", "sem DOI"))
    list_results.append((titulo, link,abstract))
```

```python
list_results[0]
```

```python
pdf = FPDF()
pdf.set_auto_page_break(auto=True, margin=15)
pdf.add_page()
pdf.set_font("Arial", size=12)

for i, (title, link, abstract) in enumerate(list_results, 1):
    pdf.set_font("Arial", style="B", size=12)
    pdf.multi_cell(0, 10, f"{i}. {title}")
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, f"Link: {link}")
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, abstract)
    pdf.ln()

pdf.output("resumos_arxiv.pdf")
```

### Openalex
O OpenAlex oferece uma alternativa aberta para bases de conhecimento científico padrão da indústria, como o Scopus da Elsevier e o Web of Science da Clarivate. Comparado a esses serviços pagos, o OpenAlex oferece vantagens significativas em termos de inclusão, acessibilidade e disponibilidade.

link: https://docs.openalex.org/how-to-use-the-api/api-overview

```python
url = "https://api.openalex.org/works"
params = {
    "filter": "title.search:emotion music,open_access.is_oa:true",
    "per_page": 5
}

response = requests.get(url, params=params).json()
# print(response)
for work in response["results"]:
    print(work["display_name"], "-", work.get("doi", "sem DOI"))
```

```python
work.keys()
```

**Abstract**

```python
# o abstract vem todo separado
for work in response["results"]:
  if work["abstract_inverted_index"]:
    # print(work["display_name"], "-", work.get("doi", "sem DOI"))
    print(work["abstract_inverted_index"])
```

```python
def recuperar_abstract(inverted_index):
    if not inverted_index:
        return "Resumo não disponível."
    index_map = {}
    for word, positions in inverted_index.items():
        for pos in positions:
            index_map[pos] = word
    return ' '.join(index_map[i] for i in sorted(index_map))

for work in response["results"]:
  if work["abstract_inverted_index"]:
    # print(work["display_name"], "-", work.get("doi", "sem DOI"))
    print(recuperar_abstract(work["abstract_inverted_index"]))
```

**Gerar PDF dos resumos**

```python
work['open_access'].get('oa_url') or work['open_access'].get('url')
```

```python
pdf_link = work['open_access'].get('oa_url') or work['open_access'].get('url')
titulo = work["display_name"]
response_link = requests.get(pdf_link)
with open(f"abstract_{titulo}.pdf", "wb") as f:
    f.write(response_link.content)
```

```python
list_results = list()
for work in response["results"]:
  titulo = work["display_name"]
  link = work.get("doi", "None")
  if work["abstract_inverted_index"]:
    # print(work["display_name"], "-", work.get("doi", "sem DOI"))
    list_results.append((titulo, link, recuperar_abstract(work["abstract_inverted_index"])))
```

```python
pdf = FPDF()
pdf.set_auto_page_break(auto=True, margin=15)
pdf.add_page()
pdf.set_font("Arial", size=12)

for i, (title, link, abstract) in enumerate(list_results, 1):
    pdf.set_font("Arial", style="B", size=12)
    pdf.multi_cell(0, 10, f"{i}. {title}")
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, f"Link: {link}")
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, limpar_unicode(abstract))
    pdf.ln()

pdf.output("resumos_openalex.pdf")
```
